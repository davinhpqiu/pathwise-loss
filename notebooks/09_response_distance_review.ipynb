## Definitions

Real matrix tuples have joint budget $|A|^2=\sum_i\lVert A_i\rVert_F^2\le r^2$.
Let $R_A$ evaluate tensor words as chronological matrix products.
For full decoded series $F,G$, define

$$
\Delta_{r,n}(F,G)=\sup_{|A|\le r}
\lVert R_A(F)-R_A(G)\rVert_{\mathrm{op}},\qquad
\mathcal B_r(F,G)=\sum_{j\ge1}r^j\lVert F_j-G_j\rVert_{\mathrm{HS}}.
$$

The output norm is Euclidean operator norm; tensors use canonical HS norm.
One tuple acts across all levels and intervals. Full HS certificate,
additional length envelope and true response are distinct constraints.


## Core result: ordinary planar paths

For a continuous planar BV path $X$ and a fixed partition $\pi$, with local
homogeneous log-signatures $\ell_{i,j}=(\log_\otimes S(X_i))_j$ on interval
$X_i$, compare

$$
F=S(X),\qquad G_\pi=\prod_i\exp_\otimes(\ell_{i,1}+\ell_{i,2}).
$$

Dilation $\delta_s$ multiplies degree $j$ by $s^j$.
The first discrepancy is $D_3=\sum_i\ell_{i,3}$. If
$h=\lVert D_3\rVert_{\mathrm{HS}}>0$, then under path/coefficient joint scale
$t=sr\downarrow0$,

$$
\frac{\Delta_{r,n}(\delta_sF,\delta_sG_\pi)}
{\mathcal B_r(\delta_sF,\delta_sG_\pi)}
\longrightarrow\rho=\frac{2\sqrt2}{9},\qquad n\ge2,
$$

uniformly in response dimension. Notebook 08 gives positive higher-degree
tails, an explicit finite-scale rejection gap and BV perturbation radii.
It also covers mixed displacement/area blocks.

The partition and cubic nondegeneracy are essential. A square loop can have
zero cubic discrepancy but nonzero quartic error. Mesh refinement can make
$h$ vanish; mesh-uniform conclusions and rough input with $2<p<3$ remain open.


## Constants, dimensions and homogeneous merges

Define $\gamma_{k,n}(D)=\sup_{|A|\le1}\lVert R_A(D)\rVert_{\mathrm{op}}$
and $\rho_{k,n}(D)=\gamma_{k,n}(D)/\lVert D\rVert_{\mathrm{HS}}$ for $D\ne0$.
For $E_k=\operatorname{ad}_{e_1}^{k-1}(e_2)$,

$$
\gamma_{k,n}(E_k)=\frac{[2(k-1)]^{(k-1)/2}}{k^{k/2}},\qquad
\lVert E_k\rVert_{\mathrm{HS}}=\sqrt{\binom{2k-2}{k-1}},\qquad n\ge2.
$$

The HS/response factor grows exponentially with degree. The normalized
tolerance interval has width tending to one.

| result | established internally | remaining boundary |
|---|---|---|
| planar degree two/three | ratios $1/2$ and $2\sqrt2/9$ | arbitrary higher-degree directions |
| mixed $G_4=[e_1,[e_2,[e_1,e_2]]]$ | $\rho_{4,2}=\sqrt{(411+41\sqrt{41})/20480}$ | upper bound at $n\ge3$ |
| dimension blindness | classical planar degree-six identity; nonzero $n=3$ witness | pure example requires $p\ge6$; BV example unproved |
| homogeneous merging | exact Gram coefficients; all-scale coordinate-loop chains | general infinite coefficient condition requires analytic verification |
| area families | exact individual gauge; spectral aggregate upper bounds | aggregate attainment and conservatism |
| tangent/locality | uniform exponential lemma; exact loop-order example | broad geometric or mesh-uniform conclusions |

Decoded-factor comparisons require original-path error accounting when fine
factors already approximate a path. The exact finite-scale merge response
supremum remains open. Ascent supplies lower estimates only.


## Storage

For degree-$N$ Euler decoding on $m$ intervals, put
$P_N(x)=\sum_{j=0}^Nx^j/j!$. Exact BV minimax error, allowing path-dependent
partitions, is

$$
\mathcal E(N,m,r)=e^r-P_N(r/m)^m.
$$

Equal variation gives the universal upper bound; a scalar straight line
attains it. Minimizing $m\Lambda_d(N)$ uses the same dense generic storage,
where $\Lambda_d(N)$ is cumulative free Lie dimension.
Finite integer optimum sets can move, while leading asymptotics persist.

Pure Engel response costs are bracketed by analytic lower/upper envelopes;
HS-certified overhead grows in the amplitude regime within this Euler workflow.
Pure paths also admit exact one-factor log-ODE decoding.
The final cell reproduces a finite BV optimum comparison.


## Prior work and status

- [Chevyrev--Lyons](https://arxiv.org/html/1307.3580v6), §4.2/Theorem 4.8:
  polynomial identities and development separation.
- [Nikolaev](https://www.math.bas.bg/pliska/Pliska-08/Pliska-08-1986-065-076.pdf),
  Theorem 1: our classical degree-six two-variable identity.
- [Böttcher--Wenzel](https://doi.org/10.1016/j.laa.2008.05.020):
  sharp commutator bound; Engel formula is a corollary.
- [Boedihardjo--Geng--Souris](https://arxiv.org/html/1811.12170v2), §4.2.4,
  and [Li--Gao](https://arxiv.org/html/2606.30055v1), Theorem 4.6:
  adjacent spectral/tail questions with different budgets.
- [Iserles--Kropielnicka](https://arxiv.org/html/2401.06635v1), Theorem 2.1:
  commutator error representation.
- [FPL](https://arxiv.org/html/2607.26281v1), (1.13) and after (2.10):
  scalar Euler example and leading discarded log-signature error.
- [Bayer--Breneis--Lyons](https://arxiv.org/html/2307.12590v1), Theorem 3.1:
  fixed-system adaptive error analysis.

The general machinery has precedent. Model reduction has different
quantifiers; Dirac-law PCF distances already compare deterministic paths.
A linear Frobenius budget does not certify nonlinear neural CDE errors.

Internal checks include exact $G_4$ elimination, concavity coefficients,
tensor expansions, explicit developments and high-precision scalar anchors.
External mathematical reading and exact-ratio novelty verification remain
outstanding. Current priority is validation of cubic/tail/stability and
all-scale proofs; the rough-control extension is a subsequent target.
